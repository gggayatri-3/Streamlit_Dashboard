```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("D:\\Home_credit\\raw_datasets\\application_train.csv")
df.head()
```

```python
required_columns = [
    "SK_ID_CURR",
    "DAYS_BIRTH",
    "DAYS_EMPLOYED",
    "AMT_INCOME_TOTAL",
    "AMT_CREDIT",
    "AMT_ANNUITY",
    "AMT_GOODS_PRICE",
    "CNT_FAM_MEMBERS"
]
df[required_columns].head()
```

```python
df["AGE_YEARS"] = (-df["DAYS_BIRTH"] / 365.25).round(1)
```

```python
df[["DAYS_BIRTH", "AGE_YEARS"]].head()
```

```python
df["AGE_YEARS"].describe()
```

```python
df["EMPLOYMENT_YEARS"] = np.where(
    df["DAYS_EMPLOYED"] == 365243,
    np.nan,
    (-df["DAYS_EMPLOYED"] / 365.25).round(1)
)
```

```python
df[["DAYS_EMPLOYED", "EMPLOYMENT_YEARS"]].head(10)
```

```python
df["EMPLOYMENT_YEARS"].describe()
```

```python
df["EMPLOYMENT_YEARS"].isnull().sum()
```

```python
df["AGE_GROUP"] = pd.cut(
    df["AGE_YEARS"],
    bins=[0, 30, 40, 50, 60, np.inf],
    labels=["20–30", "31–40", "41–50", "51–60", "60+"],
    include_lowest=True
)
```

```python
df["AGE_GROUP"].value_counts().sort_index()
```

```python
df[["AGE_YEARS", "AGE_GROUP"]].head(10)
```

```python
# Create EMPLOYMENT_GROUP
df["EMPLOYMENT_GROUP"] = pd.cut(
    df["EMPLOYMENT_YEARS"],
    bins=[-np.inf, 1, 3, 5, 10, 20, np.inf],
    labels=[
        "<1 Year",
        "1–3 Years",
        "3–5 Years",
        "5–10 Years",
        "10–20 Years",
        "20+ Years"
    ],
    right=False
)

df["EMPLOYMENT_GROUP"] = df["EMPLOYMENT_GROUP"].cat.add_categories(
    ["Unemployed / Special"]
)

df["EMPLOYMENT_GROUP"] = df["EMPLOYMENT_GROUP"].fillna(
    "Unemployed / Special"
)
```

```python
df["EMPLOYMENT_GROUP"].value_counts()
```

```python
df[["EMPLOYMENT_YEARS", "EMPLOYMENT_GROUP"]].head(10)
```

```python
df["INCOME_GROUP"] = pd.qcut(
    df["AMT_INCOME_TOTAL"],
    q=5,
    labels=["Very Low", "Low", "Middle", "High", "Very High"],
    duplicates="drop"
)
```

```python
df["INCOME_GROUP"].value_counts().sort_index()
```

```python
df[["AMT_INCOME_TOTAL", "INCOME_GROUP"]].head(10)
```

```python
# Create Affordability Ratios
df["CREDIT_TO_INCOME"] = (
    df["AMT_CREDIT"] / df["AMT_INCOME_TOTAL"].replace(0, np.nan)
)

df["ANNUITY_TO_INCOME"] = (
    df["AMT_ANNUITY"] / df["AMT_INCOME_TOTAL"].replace(0, np.nan)
)

df["GOODS_TO_INCOME"] = (
    df["AMT_GOODS_PRICE"] / df["AMT_INCOME_TOTAL"].replace(0, np.nan)
)

df["CREDIT_TO_GOODS"] = (
    df["AMT_CREDIT"] / df["AMT_GOODS_PRICE"].replace(0, np.nan)
)
```

```python
df[
    [
        "AMT_INCOME_TOTAL",
        "AMT_CREDIT",
        "AMT_ANNUITY",
        "AMT_GOODS_PRICE",
        "CREDIT_TO_INCOME",
        "ANNUITY_TO_INCOME",
        "GOODS_TO_INCOME",
        "CREDIT_TO_GOODS"
    ]
].head()
```

```python
df[
    [
        "CREDIT_TO_INCOME",
        "ANNUITY_TO_INCOME",
        "GOODS_TO_INCOME",
        "CREDIT_TO_GOODS"
    ]
].describe()
```

```python
df["INCOME_PER_FAMILY_MEMBER"] = (
    df["AMT_INCOME_TOTAL"] /
    df["CNT_FAM_MEMBERS"].replace(0, np.nan)
)
```

```python
df[
    [
        "AMT_INCOME_TOTAL",
        "CNT_FAM_MEMBERS",
        "INCOME_PER_FAMILY_MEMBER"
    ]
].head(10)
```

```python
df["INCOME_PER_FAMILY_MEMBER"].describe()
```
